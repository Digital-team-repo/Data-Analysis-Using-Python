**Objective:** Use a simple temperature example to understand the IPO workflow (Input, Process, Output) 

**Dataset description:** This file shows annual average temperature departure values for Canada (difference from baseline).

**Data source:** This version uses the cleaned CSV file (data_en.csv) located in this folder, adapted from Environment and Climate Change Canada (ECCC).

## Step 1 (Process Setup): Import libraries

Big picture: We load pandas for table operations and plotly for visualization.

Technical details: pandas creates a DataFrame named df. plotly.express builds the chart.

```python
import pandas as pd
import plotly.express as px
```

## Step 2 (Input): Load the data file

Big picture: We open the CSV file and store it as a table.

Technical details: pd.read_csv() reads data_en.csv into df.

```python
data_url = "https://raw.githubusercontent.com/Digital-team-repo/Data-Analysis-Using-Python/refs/heads/main/2-data-analysis-intro/data_en.csv"

df = pd.read_csv(data_url)
df.head()
```

## Step 3 (Process): Inspect columns

Big picture: We check column names before creating a chart.

Technical details: df.columns shows column labels.

```python
df.columns
```

## Step 4 (Output): Create a visualization

Big picture: We plot Year vs Departure to show changes over time.

Technical details: px.scatter() draws points using specified x and y columns.

```python
fig = px.scatter(
    df, 
    x="Year", 
    y="Temperature departure (degree Celsius)",
    title="Annual Temperature Change in Canada (1948-2024)",
    trendline="ols"
)

fig.show()
```
