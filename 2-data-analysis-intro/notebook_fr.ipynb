**Objectif :** L’exemple des températures qui suit illustre le flux de travail ETS–entrée, traitement, sortie–tel qu’il se déploie dans un carnet Jupyter.

**Description du jeu de données :** Ce fichier consigne les écarts annuels moyens de température au Canada (la différence par rapport à la normale de 1961 à 1990) de 1948 à 2024.

**Source des données :** Cette version s’appuie sur le **fichier CSV français épuré** (`data_fr.csv`) situé dans ce dossier. Il s’agit d’une adaptation du jeu de données original accessible sur la page de l’indicateur Changements de la température du gouvernement du Canada.


## Étape 1 (configuration du traitement) : Importer les bibliothèques

Cette étape de configuration s’exécute en premier afin que nous puissions ensuite passer à l’étape des entrées (`pd.read_csv(...)`), puis poursuivre le traitement.


```python
import pandas as pd
import plotly.express as px
```

## Étape 2 (entrée) : Charger le fichier de données

Nous chargeons `data_fr.csv` depuis ce dossier comme entrée de notre flux de travail.


```python
data_url = "https://raw.githubusercontent.com/Digital-team-repo/Data-Analysis-Using-Python/refs/heads/main/2-data-analysis-intro/data_fr.csv"
df = pd.read_csv(data_url)
df.head()
```

## Étape 3 (traitement) : Inspecter les colonnes

Nous vérifions le nom des variables avant de tracer le graphique afin que l’étape de traitement soit intentionnelle.


```python
df.columns
```

## Étape 4 (sortie) : Créer une visualisation

Ce graphique constitue notre sortie : une synthèse visuelle de la tendance des données au fil du temps.


```python
fig = px.scatter(
    df,
    x="Année",
    y="Écart de températures (degrés Celsius)",
    title="Changement des températures annuelles au Canada (1948-2024)",
    trendline="ols"
)

fig.show()
```
