## Step 1: Import pandas
Big picture: We use pandas to load and clean table data.

```python
import pandas as pd
```

## Example 1: Missing values and duplicates
Big picture: This file looks normal, but it includes blanks and duplicate rows.

```python
data1_url = "https://raw.githubusercontent.com/Digital-team-repo/Data-Analysis-Using-Python/main/05-data-cleaning/data1_en.csv"
df1 = pd.read_csv(data1_url)
df1.head()
```

```python
df1.isna().sum()
```

```python
df1_clean = df1.copy()
df1_clean = df1_clean.dropna(subset=["Total Students", "Program"])
df1_clean = df1_clean.drop_duplicates()
len(df1), len(df1_clean)
```

## Example 2: Extra lines before the header
Big picture: Some open-data files include title/notes lines before the real table header.

```python
data2_url = "https://raw.githubusercontent.com/Digital-team-repo/Data-Analysis-Using-Python/main/05-data-cleaning/data2_en.csv"
# skiprows=4 tells pandas to ignore the first 4 non-data lines
df2 = pd.read_csv(data2_url, skiprows=4)
df2.head()
```

```python
df2_clean = df2.copy()
df2_clean = df2_clean.dropna(subset=["Total Students"])
df2_clean = df2_clean.drop_duplicates()
len(df2), len(df2_clean)
```

## Example 3: Inconsistent text and non-numeric values
Big picture: Here we clean spaces/case and convert numbers safely.

```python
data3_url = "https://raw.githubusercontent.com/Digital-team-repo/Data-Analysis-Using-Python/main/05-data-cleaning/data3_en.csv"
df3 = pd.read_csv(data3_url)
df3.head()
```

```python
df3_clean = df3.copy()
df3_clean["Region"] = df3_clean["Region"].fillna("Unknown").str.strip().str.title()
df3_clean["Total Students"] = pd.to_numeric(df3_clean["Total Students"], errors="coerce")
df3_clean = df3_clean.dropna(subset=["Total Students", "Program"])
df3_clean = df3_clean.drop_duplicates()
df3_clean.head()
```

```python
len(df3), len(df3_clean)
```
