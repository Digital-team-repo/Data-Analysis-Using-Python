- **Code cell:** a notebook cell that stores and runs Python code
- **Kernel:** the background Python engine running your notebook
- **Execution count:** the number next to a code cell (for example, [1]) showing when it ran
- **PEMDAS:** Python order of operations (Parentheses, Exponents, Multiplication/Division, Addition/Subtraction)
