- **Cellule de texte :** cellule de carnet réservée au texte rédigé, comme les titres, les notes et les explications
- **Cellule de code :** cellule de carnet qui exécute du code Python
- **Zone de sortie :** résultat visible affiché après l’exécution d’une cellule de code
- **Noyau ou environnement d’exécution :** moteur de fond qui prend en charge l’exécution du code d’un carnet
- **PEMDAS :** ordre des opérations appliqué par Python lors de l’évaluation des expressions
