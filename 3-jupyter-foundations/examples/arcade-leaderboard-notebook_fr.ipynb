## Étape 1 : Importer les outils


```python
import random
from statistics import mean
import matplotlib.pyplot as plt
```

## Étape 2 : Créer un échantillon de données sur la classe


```python
players = ["Avery", "Jordan", "Kai", "Mina", "Riley", "Sasha"]
scores = [random.randint(1200, 3800) for _ in players]
bonus_round = [random.randint(-300, 500) for _ in players]
final_scores = [s + b for s, b in zip(scores, bonus_round)]
```

## Étape 3 : Classer et résumer les résultats


```python
ranked = sorted(zip(players, final_scores), key=lambda x: x[1], reverse=True)
ranked
```

```python
print("Classement de la soirée arcade")
for i, (name, score) in enumerate(ranked, start=1):
    print(f"{i}. {name} : {score}")

print(f"\nScore moyen de la classe : {mean(final_scores):.1f}")
print(f"Écart entre le meilleur score et le plus faible : {ranked[0][1] - ranked[-1][1]}")
```

## Étape 4 : Visualiser le classement


```python
names = [n for n, _ in ranked]
values = [v for _, v in ranked]
colors = ["#1f77b4" if i == 0 else "#8fb9e3" for i in range(len(values))]

plt.figure(figsize=(8, 4))
plt.bar(names, values, color=colors)
plt.title("Scores finaux de la soirée arcade")
plt.xlabel("Joueur")
plt.ylabel("Score final")
plt.grid(axis="y", linestyle="--", alpha=0.7)
plt.show()
```
