## Vocabulaire de base

- **`import` :** commande Python servant à intégrer une bibliothèque dans le carnet
- **pandas (`pd`) :** bibliothèque Python pour travailler avec des données tabulaires
- **`data_url` :** variable qui stocke le lien vers le fichier CSV avant son chargement
- **`pd.read_csv(data_url)` :** commande qui charge les données CSV dans un tableau appelé DataFrame
- **DataFrame (`df`) :** tableau créé après le chargement des données avec pandas
- **`df.columns` :** propriété d’un DataFrame qui affiche le nom des colonnes
- **`df.head()` :** méthode qui affiche les premières lignes
- **`df.tail()` :** méthode qui affiche les dernières lignes
