## Key Vocabulary

- **import**: a Python command used to bring a library into the notebook (for example import pandas as pd).
- **pandas (pd)**: a Python library for working with table-shaped data.
- **data_url**: a variable name used to store the link to the CSV file before loading.
- **pd.read_csv(data_url)**: a command that loads CSV data into a table (DataFrame).
- **DataFrame (df)**: tThe table object created after loading data with pandas.
- **df.columns**: a DataFrame property that shows the column names in the table.
- **df.head()**: shows the first rows.
- **df.tail()**: shows the last rows.

## Step 1: Import libraries

## Step 2: Open and Load the CSV

Load the CSV file into a DataFrame variable named df.

```python
data_url = "https://raw.githubusercontent.com/Digital-team-repo/Data-Analysis-Using-Python/refs/heads/main/2-data-analysis-intro/data_en.csv"
df = pd.read_csv(data_url)
```
