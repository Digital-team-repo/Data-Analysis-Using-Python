## Step 1: Set the JSON URL
Big picture: We store the JSON link in one variable so it is easy to change later.

```python
data_url = "https://open.canada.ca/data/en/api/3/action/package_show?id=d38de914-c94c-429b-8ab1-8776c31643e3"
```

## Step 2: Import pandas
Big picture: We use pandas to read the JSON into a table-like structure.

```python
import pandas as pd
```

## Step 3: Read the JSON
Big picture: This loads the JSON from `data_url` into `df`.

```python
df = pd.read_json(data_url)
df.head()
```
