## Étape 1 : Importer les bibliothèques

Vue d’ensemble : nous chargeons les outils nécessaires à l’ouverture des données Excel et à la création d’un graphique.

Détails techniques : `pandas` crée un DataFrame nommé `df`. `plotly.express` construit le graphique.


```python
import pandas as pd
import plotly.express as px
```

## Étape 2 : Charger le fichier Excel

Vue d’ensemble : nous ouvrons le fichier Excel français prêt à l’utilisation en classe, stocké dans ce même dossier.

Détails techniques : `pd.read_excel()` lit la feuille de calcul et la charge dans `df`.


```python
data_url = "https://github.com/Digital-team-repo/Data-Analysis-Using-Python/raw/refs/heads/main/5-excel-analysis/data_fr.xlsx"
df = pd.read_excel(data_url)
```

## Étape 3 : Vérifier le tableau

Vue d’ensemble : nous inspectons la structure et le nombre de lignes avant de créer des graphiques.

Détails techniques : `df.columns` et `df.shape` sont des propriétés, que l’on lit sans parenthèses. `df.head(10)` et `df.tail(10)` sont des méthodes, que l’on exécute avec des parenthèses. `df.shape` montre la taille du tableau sous la forme `(lignes, colonnes)`, tandis que `len(df)` renvoie seulement le nombre de lignes.


```python
print(df.columns)
```

```python
print(df.shape)
```

```python
print(len(df))
```

```python
print(df.head(10))
```

```python
print(df.tail(10))
```

## Étape 4 : Créer un histogramme pour toutes les écoles

Vue d’ensemble : nous représentons d’abord l’intégralité du jeu de données afin d’établir une vue de référence de la répartition des effectifs.

Détails techniques : un histogramme regroupe les effectifs en intervalles et indique le nombre d’écoles dans chaque tranche.


```python
fig_all = px.histogram(
    df,
    x="l’effectif",
    nbins=30,
    title="Répartition des effectifs scolaires (toutes les écoles)"
)

fig_all.show()
```

## Étape 5 : Filtrer les écoles secondaires (étape du traitement)

Vue d’ensemble : nous affinons maintenant le jeu de données pour répondre à une question plus précise.

Détails techniques : nous appliquons un filtrage booléen avec `df[df["le niveau de l’école"] == "Secondaire"]` et stockons les résultats dans `secondary_df`.


```python
secondary_df = df[df["le niveau de l’école"] == "Secondaire"].copy()
print(secondary_df.shape)
print(len(secondary_df))
print(secondary_df.head(10))
print(secondary_df.tail(10))
```

## Étape 6 : Créer un histogramme limité aux écoles secondaires

Vue d’ensemble : nous comparons ce deuxième graphique à celui de toutes les écoles pour voir comment le filtrage modifie la répartition.

Détails techniques : cet histogramme utilise la même colonne `l’effectif`, mais seulement à partir des données de `secondary_df`.


```python
fig_secondary = px.histogram(
    secondary_df,
    x="l’effectif",
    nbins=30,
    title="Répartition des effectifs scolaires (écoles secondaires seulement)"
)

fig_secondary.show()
```
