## Key Vocabulary

- **NaN:** stands for Not a Number and usually means a value is missing or invalid in your dataset
- **Duplicate row:** a row that appears more than once in a dataset
- **Metadata line:** a non-data line in a file (for example, a title or note) that appears before the real header
- **`skiprows`:** an option in file-reading commands (like `pd.read_csv`) that skips a specified number of lines
