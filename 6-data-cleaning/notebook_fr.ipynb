## Vocabulaire de base

- **`NaN` :** abréviation de *Not a Number* qui indique habituellement qu’une valeur est manquante ou invalide dans le jeu de données
- **Ligne en double :** ligne qui apparaît plus d’une fois dans un jeu de données
- **Ligne de métadonnées :** ligne placée au début d’un fichier et ne faisant pas partie du tableau
- **`skiprows` :** option de pandas qui ignore les premières lignes du tableau avant la lecture
